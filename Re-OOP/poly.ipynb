```python
############################# Method Overloading #############

class Geometry:

    # Python does not support traditional method overloading (same method name with different signatures).
    # Instead, we achieve method overloading using default arguments (e.g., b=0).
    def area(self, a, b=0):
        if b == 0:
            print("Circle", 3.14 * a * a)
        else:
            print("Rect", a * b)

obj = Geometry()

# Passing 1 argument -> b defaults to 0 -> calculates Circle area
obj.area(4)

# Passing 2 arguments -> b takes value 5 -> calculates Rectangle area
obj.area(4, 5)
```
